```python
import pandas as pd

data_set = pd.read_csv('../Data/student_performance.csv')
data_set.head(1)
```

```python
data_set.isnull().sum()
```

```python
new_data_set = data_set[['study_hours','exam_score']]
new_data_set
```

```python
new_data_set['study_hours'].corr(new_data_set['exam_score'])
```

```python
import matplotlib.pyplot as plt

plt.scatter(new_data_set['study_hours'], new_data_set['exam_score'])
plt.xlabel("Study Hour")
plt.ylabel("Study Score")

plt.show()
```

```python
X = new_data_set[['study_hours']]
Y = new_data_set['exam_score']

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, Y_train)

prediction = model.predict(pd.DataFrame({'study_hours': [10]}))
prediction
```
