```python
#check the outliners
numerical_data = row_data_set.select_dtypes("number").columns

Q1 = row_data_set[numerical_data].quantile(0.25)
Q3 = row_data_set[numerical_data].quantile(0.75)

IQR = Q3 - Q1

outlier_mask = (
    row_data_set[numerical_data] < (Q1 - 1.5 * IQR)) | (row_data_set[numerical_data] > (Q3 + 1.5 * IQR)
)

all_outlier_mask = row_data_set[outlier_mask.any(axis=1)]
all_outlier_mask.shape
```
