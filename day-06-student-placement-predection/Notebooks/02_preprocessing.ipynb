```python
import pandas as pd

row_data_set = pd.read_csv("../Data/Row/student_placement_Data.csv")
row_data_set.info()
```

```python
# remove outliers
all_numeric_data_set = row_data_set.select_dtypes("number").columns

Q1 = row_data_set[all_numeric_data_set].quantile(0.25)
Q3 = row_data_set[all_numeric_data_set].quantile(0.75)

IQR = Q3 - Q1

outlier_mask = (
    (row_data_set[all_numeric_data_set] < (Q1 - 1.5 * IQR))
    | (row_data_set[all_numeric_data_set] > (Q3 + 1.5 * IQR))
)

all_outliers = row_data_set[outlier_mask.any(axis=1)]

row_data_set.drop(index=all_outliers.index, inplace=True)
row_data_set.shape
```

```python
row_data_set['salary_package_lpa'].isnull().sum()
```

```python
row_data_set['salary_package_lpa'] = row_data_set['salary_package_lpa'].fillna(0).astype(int)
row_data_set.head(5)
```

```python
all_categorical_data_types = row_data_set.select_dtypes("string").columns
all_categorical_data_types
```

```python
row_data_set["branch_code"] = row_data_set["branch"].astype("category").cat.codes
row_data_set["college_tier_code"] = row_data_set["college_tier"].astype("category").cat.codes

row_data_set.head(3)
```

```python
row_data_set.to_csv("../Data/Processed/processed_student_data_set.csv")
new_data_set = pd.read_csv("../Data/Processed/processed_student_data_set.csv")
new_data_set.isnull().sum()
```
