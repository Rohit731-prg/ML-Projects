```python
import pandas as pd

data_set = pd.read_csv("student_data_set.csv")
data_set.head(5)
```

```python
data_set.isnull().sum()
```

```python
X = data_set[['Study Hours']]
Y = data_set['Passed']

import matplotlib.pyplot as plt

plt.scatter(x=X, y=Y)
plt.xlabel("Study Hour")
plt.ylabel("Student pass/not")
plt.title("Student data")

plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.3, random_state=42
)

plt.scatter(X_test, Y_test)
plt.xlabel("Student_study_hour")
plt.ylabel("Student_pass_not")

plt.title("Student data in test dataset")
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train, Y_train)

user_input = int(input("Enter user study hour: "))
predict = model.predict(pd.DataFrame({"Study Hours": [user_input]}))
predict
```

```python
coefficient = model.coef_
intercept = model.intercept_
print("coefficient: ", coefficient)
print("intercept: ", intercept)
```
