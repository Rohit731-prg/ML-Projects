```python
X = data_set[['Training_Hours_Per_Week']]
Y = data_set['Win_Rate_Pct']
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, Y)
plt.xlabel("Training_Hours_Per_Week")
plt.ylabel("Win_Rate_Pct")

plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

X_train, Y_train
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X_train, Y_train)
```

```python
user_input = int(input("Enter your value: "))
prediction = model.predict(pd.DataFrame({"Training_Hours_Per_Week": [user_input]}))
prediction
```
